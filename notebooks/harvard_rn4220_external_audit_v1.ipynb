```python
from pathlib import Path
import json, gzip, hashlib
root = Path('/Users/tt/Documents/ChatGPT/Train an antibiotic optimization model')
folder = root / "data/processed/harvard_rn4220_external_v1"
manifest = json.loads((folder / "manifest.json").read_text())
manifest["inventories"]
```

```python
for name, expected in manifest["output_hashes"].items():
    assert hashlib.sha256((folder / name).read_bytes()).hexdigest() == expected
verification = json.loads((root / "reports/harvard_rn4220_external_verification_v1.json").read_text())
assert verification["status"] == "passed"
verification
```
